# What is broadcasting?

# stretching the smaller array to the shape of the bigger one

When two arrays of different shapes meet in an operation, NumPy repeats the smaller one along the missing dimension - a number for every element, a row for every row - so no loop is needed. Shapes are compared from the right; a dimension of size 1 (or missing) is stretched, anything else is an error.

`arr * 4` - the number is stretched to (3, 3); `iris - iris.mean(axis=0)` - the row of 4 means is repeated for all 150 rows
