## Functions

If you anticipate needing to repeat the same or very similar code more than once, it may be worth writing a reusable function.

You already **use** functions: `print()`, `len()`, `input()`, `range()`. Today we learn to **write our own**.

A function works like a machine: arguments go in (11), the function applies its steps (x + 1), and a return value comes out (12)
function_machine.png


### Functions

* **parameter** - the name inside the brackets in `def` (here `x`)
* **argument** - the real value you pass when you call the function (here `11`)
* the body is **indented** (4 spaces), as in `if` and `for`
* `def` only *describes* the function. Nothing is executed until you **call** it: `plusone(11)`

````python
def plusone(x):
    "This is a short explanation on your function: this function adds 1 to the passed argument"
    print(f"Calling the 'plusone' function with argument '{x}'")
    return x+1
````

### Function arguments

* Positional vs. Keyword arguments
* Default values

Positional arguments are matched by order, keyword arguments by name, and a default value is used when an argument is not given
arguments.png

````python
def my_power(base, exponent):
    "Return the 'base' to the power of 'exponent'"
    return(base**exponent)

print(my_power(2,3))
print(my_power(3,2))
````

````python
def my_power(base, exponent=1):
    "Return the 'base' to the power of 'exponent'"
    return(base**exponent)

print(my_power(2,3))
print(my_power(3,2))
print(my_power(base=3,exponent=2))
print(my_power(exponent=2,base=3))
print(my_power(5))        # exponent is not given: the default value 1 is used
````

### Function arguments

* Multiple positional arguments

`*names` collects **any number** of values into a **tuple** (useful when you do not know in advance how many values will come).

````python
def helloAll1(*names):
    for name in names:
        print(f"Hello, {name}!")

helloAll1("John", "Jim")
````

* Multiple keyword arguments

`**namesDict` collects any number of **named** values into a **dictionary**. (useful when  the number of named arguments may vary.)

````python
def helloAll2(**namesDict):
    for name in namesDict:
        print(f"Hello, {name} {namesDict[name]}!")

    #print(namesDict)

helloAll2(Mr = "John", Dr = "Jim")
````

* All types of arguments (useful when a function needs required inputs plus a flexible number of extra inputs)

The order is always: normal arguments, then `*args`, then `**kwargs`.

````python
def helloAll3(first, *names, **namesDict):
    print(f"Hello, {first}!!!!!")

    for name in names:
        print(f"Hello, {name}!")
    for name in namesDict:
        print(f"Hello, {name} {namesDict[name]}!")


helloAll3("Marta", "Jane", "Jack", Mr = "John", Dr = "Jim")
````

Let us see what exactly is collected:

````python
def show(*a, **k):
    print(type(a), a)
    print(type(k), k)

show(1, 2, 3, city="Riga", year=2026)
````

## Functions are objects

Functions are the objects, so you can pass them as arguments.

A function name **without brackets** is the function itself (like any other value); with brackets `()` the function is **called**.

````python
def say_hello():
    return "Hello!"

f = say_hello          # gives the same function another name; refering function
print(f())             # the function is called
````

A function can be an argument of another function. Example: clean the list of names by applying a **list of functions** one after another:

1. `str.strip` - removes spaces around the text
2. `remove_punctuation` - removes the symbols `! # ? .`
3. `str.capitalize` - first letter big, the others small

````python
students = ('aLexey!', '    anar', '..erika', ' Jevgenijs!!! ', '##janis', 'vjaceSLAV', 'NIKITa')

import re
def remove_punctuation(value):
    return re.sub("[!#?.]", "", value)


def run_all_funcs(strings, funcs):  # applies a sequence of functions to every item and returns the results in a list
    result = []
    for value in strings:
        for func in funcs:
            value = func(value)
        result.append(value)
    return result

clean_funcs = [str.strip, remove_punctuation, str.capitalize]   # Create and choose the cleaning functions and their order

students_cleaned = run_all_funcs(students,clean_funcs)

print(students_cleaned)
````

## Namespaces

A namespace is a 'place' with a set of names and their corresponding objects. Think of two different classrooms that both have a student called Anna: when the teacher says "Anna" inside a classroom, the Anna from **this** classroom answers.

Three nested namespaces: built-in, global and local. Python looks for a name first in the local namespace, then in the global one, and last in the built-in one
namespaces.png

## Namespaces

* The built-in namespace contains the names of all of Python's built-in objects. These are available at all times when Python is running.
* The global namespace contains any names defined at the level of the main program. Python creates the global namespace when the main program body starts, and it remains in existence until the interpreter terminates.
* Python creates a new namespace whenever a function executes. That namespace is local to the function and remains in existence until the function terminates.

````python
val = "My Value"
def func():
    val = "My Local Value"
    print("Local value: " + val)

func()
print(val)
````

### Globals and Locals dictionaries

* globals() returns an actual reference to the dictionary that contains the global namespace
* locals() returns a dictionary that is a current copy of the local namespace, not a reference to it.

````python
val = "My Value"
def func():
    val = "My Local Value"
    local_vars = locals()
    print("Local value: " + local_vars["val"])
    global_vars = globals()
    print("Global value: " + global_vars["val"])

func()
print(val)
````

The keyword `global` lets a function change a global variable:

````python
val = "My Value"
def func():
    global val
    val = "My Local Value"

func()
print(val)
````

## Lambda functions

Python supports _anonymous_ or _lambda_ functions, which are a way of writing functions consisting of a single statement, the result of which is the return value.

It's often clearer to pass a lambda function as an argument as opposed to writing a complete function declaration:

```
lambda  argument :  result
```

````python
def square(x):
    return x * x

square_lambda = lambda x: x * x      # the same function in one line
print(square(4), square_lambda(4))
````

````python
print(students_cleaned)

funcs = [lambda s: s+"! ",
        lambda s: s*3,]

run_all_funcs(students_cleaned,funcs)
````

## Libraries

Python alone knows only the basics. **Libraries** are toolboxes written by other people - you import them and use their functions.

A library is a collection of packages, a package is a folder of modules, a module is one .py file, and a module contains functions and variables
libraries.png

A __library__ is an umbrella term referring to a reusable chunk of code. It is often assumed that while a __package__ is a collection of __modules__, a __library__ is a collection of __packages__.

So
* The __module__ is a simple Python file that contains collections of functions and variables and with having a .py extension file.
* The __package__ is a simple directory having collections of modules.
* The __library__ is having a collection of packages.

## Standard libraries

* Standard (built-in) libraries are usually installed with Python
* A long list of built-in libraries https://docs.python.org/3/py-modindex.html
* Most frequently used - __math__, __random__, __statistics__, __datetime__, __json__, etc.
* Need to be imported before usage

### if \_\_name\_\_ == "\_\_main\_\_":

This statement allows distinguishing two cases:
* running the script as a separate programme (__True__ in this case)
* loading the script as a library  (__False__ in this case)

Use it for code that must run only when you start the file directly (for example tests or a demo).

If a file is run as a program, __name__ is "__main__" and the block runs; if it is imported as a library, __name__ is the module name and the block is skipped
name_main.png

**Run as a program** (the block runs):

````python
!python mylib.py
````

**Import as a library** (nothing is printed - the block is skipped). Python remembers an already imported module, so we reload it after the change:

````python
import importlib
import mylib
importlib.reload(mylib)
print("name inside the notebook:", __name__)
print("name of the module     :", mylib.__name__)
````
