# Control Flow

By default Python executes the code **line by line, from top to bottom**. Control flow statements change this order:

* __if else__ conditional statements, execute a block of code **only if** a condition is true
* __for__ loops, repeat a block of code for **each element** of a collection / a predefined number of times
* __while__ loops, repeat a block of code **as long as** a condition is true

Special statements inside loops: __break__, __continue__, __pass__

# Conditional Statements

## Comparison operators

A condition is an expression that evaluates to `True` or `False` (data type `bool`).

| Symbol | Task performed |
|---|---|
| `==` | True, if equal |
| `!=` | True, if not equal to |
| `<` | less than |
| `<=` | less than or equal to |
| `>` | greater than |
| `>=` | greater than or equal to |

Do not confuse: `=` **assigns** a value to a variable, `==` **compares** two values.

````python
# NOTE: We declare variables using '='
a = 5
b = 7
````

````python
# But compare them using '=='
print(a == b)

# Test whether a does not equal b
print(a != b)
````

````python
# Logic expressions evaluate to True or False (datatype: bool)
test = b > a
print(test)
print(type(test))
````

### Logical operators: `and`, `or`, `not`

| Operator | Result |
|---|---|
| `A and B` | True if **both** A and B are True |
| `A or B` | True if **at least one** of A, B is True |
| `not A` | the opposite of A |

* Conditions can be combined: `x > 0 and x < 10`
* Python also allows a **chained comparison**: `0 < x < 10`
* `in` / `not in` test membership in a list, tuple, set, dictionary (keys) or string

````python
x = 7

print(x > 0 and x < 10)
print(0 < x < 10)            # the same as above
print(x < 0 or x > 5)
print(not x > 5)

print(x in [1, 7, 9])
print("a" in "Data Analytics")
print("Riga" not in {"Tallinn", "Vilnius"})
````

````python
x = int(input("Insert an integer number:"))

if x > 30:
    print("You succeeded!")
elif x == 30:
    print("You are in the middle!")
else:
    print("Sorry, you failed!")

print("Thanks for joining us!")
````

### Common mistake: the order of conditions

````python
score = 95

if score >= 50:
    print("Pass")
elif score >= 90:
    print("Excellent")      # never reached! the first condition is already True for 95

# Correct: start with the most specific condition
if score >= 90:
    print("Excellent")
elif score >= 50:
    print("Pass")
else:
    print("Fail")
````

# The while loop

Keep on iterating until the condition is __False__ (initially unknown number of times)

```
while condition:
    code block
```

Three parts of a typical `while` loop:
1. **initialise** the variable used in the condition,
2. the **condition**,
3. **update** the variable inside the loop — otherwise the loop never ends (an *infinite loop*; in Colab stop it with the ■ button).

````python
x = 0
total = 0
while total < 100:
    x = x + 1
    total = total + x
    print(f"x = {x}")
print(f"Total = {total}")
````

### Counter pattern and shorthand operators

* `x += 1` is the same as `x = x + 1` (also `-=`, `*=`, `/=`)
* A **counter** counts the repetitions, an **accumulator** collects a sum

````python
countdown = 5
while countdown > 0:
    print(countdown)
    countdown -= 1
print("Go!")
````

# The for loop

Iterate a __predefined__ number of times — once for **each element** of a sequence (string, list, tuple, ...) or of another iterable (dictionary, set, `range`).

```
for element in collection:
    code block (uses element)
```

````python
for char in "Programming":
    print(char)
````

## Iterating using _range_

* `range(stop)` — numbers `0, 1, ..., stop-1`
* `range(start, stop)` — numbers from `start` to `stop-1`
* `range(start, stop, step)` — with a step (can be negative)

The `stop` value is **never** included (as in slicing).

````python
for i in range(5):
    print(f"Programming week {i}")
````

````python
for i in range(5, 15):
    print(i + i)
````

````python
print(list(range(5)))
print(list(range(2, 11, 2)))        # with a step
print(list(range(10, 0, -3)))       # counting down
````

## Iterating through a list: values and positions

* `enumerate(seq)` gives pairs `(index, element)`
* `zip(seq1, seq2)` goes through two sequences in parallel

````python
cities = ["Riga", "Daugavpils", "Liepaja", "Jelgava"]

for i, city in enumerate(cities):
    print(i, city)

print()
for i, city in enumerate(cities, start=1):       # numbering from 1
    print(f"{i}. {city}")

print()
population = [605, 78, 67, 55]                   # thousands
for city, p in zip(cities, population):
    print(f"{city}: {p}")
````

## Testing conditions inside a loop

Combining loops with conditions allows you to build more sophisticated code structures:

````python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

for day in days:
    if day == 'Sat' or day == 'Fri':
        location = 'TSI'
    elif day == 'Sun':
        location = 'Home!'
    else:
        location = 'Work + TSI'
    print(day, location)
````

## Typical loop patterns

| Pattern | Idea |
|---|---|
| **Accumulate** | start with `total = 0`, add each element |
| **Count** | start with `count = 0`, `count += 1` when a condition is True |
| **Find the largest** | remember the best value seen so far |
| **Build a new list** | start with `result = []`, `append` what you need |

(For simple cases the built-in `sum`, `len`, `max`, `min` do the job without a loop — loops are needed when the logic is more complex.)

````python
grades = [7, 9, 4, 10, 6, 8]

total = 0
count_high = 0
best = grades[0]
passed = []

for g in grades:
    total += g                   # accumulate
    if g >= 8:
        count_high += 1          # count
    if g > best:
        best = g                 # find the largest
    if g >= 5:
        passed.append(g)         # build a new list

print("Total      :", total)
print("Grades >= 8:", count_high)
print("Best       :", best)
print("Passed     :", passed)
````

## Checking the type: `isinstance`

Data from real sources is often "mixed". `isinstance(value, type)` checks whether a value has a given type; a tuple of types checks for any of them.

````python
print(isinstance(5, int))
print(isinstance("5", int))
print(isinstance(2.5, (int, float)))      # int OR float

mixed = (3, "two", 4.5, None, 10)
numbers_sum = 0
for item in mixed:
    if isinstance(item, (int, float)):
        numbers_sum += item
print("Sum of the numbers:", numbers_sum)
````

# Special statements

* __break__ jumps out of a loop
* __continue__ jumps over one iteration in the loop.


* __pass__ does nothing, but you avoid getting an error when empty code is not allowed. The statement is used as a placeholder for future code.

## Using _break_ statement

````python
for x in range(0, 5):
    if x == 3:
        break
    print(x)

print(f"The last number in the loop is {x}")
````

## Using _continue_ statement

````python
for x in range(0, 7):
    if x == 3:
        continue
    print(x)

print(f"The last number in the loop is x = {x}")
````

## Using _pass_ statement

````python
if "yellow" in {"red", "green", "blue"}:
    print("Base colour!")
else:
    pass  # Will do something later
````

### Searching with `break`

Stop as soon as the answer is found — there is no need to check the remaining elements.

````python
stops = ["Depot", "Market", "Central Station", "Old Town", "University", "Hospital"]

found = False
for stop in stops:
    if stop == "Old Town":
        found = True
        break

print("Old Town is on the route" if found else "Old Town is not on the route")
````

### Limited number of attempts: `for` + `break`

The loop is repeated at most 3 times, but can be left earlier.

````python
secret = "python"

for attempt in range(1, 4):
    answer = input(f"Attempt {attempt}/3. Password: ")
    if answer == secret:
        print("Access granted")
        break
else:
    print("Access denied")       # runs only if the loop finished WITHOUT break
````

`for ... else`: the `else` block is executed only when the loop was **not** interrupted by `break`.

### YOUR TURN 5: break and continue

`readings` are measurements of a sensor. The value `-1` means a sensor error, the value `999` means "end of transmission".

1. Calculate the sum and the number of **valid** readings, **skipping** (`continue`) the errors, and **stopping** (`break`) at the end signal. The values after `999` must not be counted.
2. Print how many readings were processed in total (valid + errors) before the loop stopped.

````python
readings = [12, 15, -1, 14, 18, -1, 13, 999, 16, 11]

# YOUR CODE HERE:
````

````python
readings = [12, 15, -1, 14, 18, -1, 13, 999, 16, 11]

total = 0
valid = 0
errors = 0

for value in readings:
    if value == 999:        # end of transmission: leave the loop
        break
    if value == -1:         # sensor error: skip to the next reading
        errors += 1
        continue
    total += value
    valid += 1

print("Sum of valid readings:", total)      # 72
print("Number of valid readings:", valid)   # 5
print("Processed in total:", valid + errors)  # 7
````

### Example: a simple quiz

A dictionary stores the questions, a loop asks them, a condition checks the answers and a counter keeps the score.

````python
import random

capitals = {"Latvia": "Riga", "Estonia": "Tallinn", "Lithuania": "Vilnius"}

score = 0
for round_no in range(1, 4):
    country = random.choice(list(capitals))
    answer = input(f"{round_no}. What is the capital of {country}? ")
    if answer.strip().lower() == capitals[country].lower():
        print("Correct!")
        score += 1
    else:
        print(f"No, it is {capitals[country]}")

print(f"Your score: {score} / 3")
````

### Example: guessing a number

The program keeps a secret number; the loop gives feedback until it is found.

````python
import random

secret = random.randint(1, 10)
guess = 0
tries = 0

while guess != secret:
    guess = int(input("Guess a number from 1 to 10: "))
    tries += 1
    if guess < secret:
        print("Too low")
    elif guess > secret:
        print("Too high")

print(f"Correct! It took {tries} tries")
````
