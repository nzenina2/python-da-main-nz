## Important Data Structures

* __Tuple__ is a fixed-length immutable sequence of Python objects, (eg., screen resolution (1920, 1080))
* __List__ is a variable-length ordered mutable sequence of Python objects, (eg., shopping items ["milk", "bread", "coffee"])
* __Set__ is an mutable unordered collection of unique Python objects, (eg., programming languages {"Python", "Java", "C++"})
* __Dictionary__ stores a mutable collection of key-value pairs, where key and value are Python objects, (eg., product data {"name": "Laptop", "price": 900}).

```python
tuple1 = (1, 2, 3)
print("Tuple ", tuple1)
list1 = [1, 2, 3]
print("List ", list1)
set1 = {1, 2, 3}
print("Set ", set1)
dict1 = {1:"One", 2:"Two", 3:"Three"}
print("Dictionary ", dict1)
```

| Structure | Ordered | Mutable | Duplicates | Access by |
|---|---|---|---|---|
| Tuple | yes | no | yes | position / index |
| List | yes | yes | yes | position / index |
| Set | no | yes | no | membership, not index |
| Dictionary | insertion order | yes | keys: no, values: yes | key |

## Indexing
Lists, tuples and strings are sequences: every element has a position (index).

* Indexing starts at 0
* Negative indices count from the end: -1 is the last element
* len(s) returns the number of elements

```python
cities = ["Riga", "Daugavpils", "Liepaja", "Jelgava", "Jurmala", "Ventspils", "Rezekne"]

print(len(cities))      # number of elements
print(cities[0])        # first element
print(cities[2])        # third element
print(cities[-1])       # last element
print(cities[-2])       # second from the end

# ERROR: list index out of range
# print(cities[7])
```

## Slicing

sequence[start:stop:step]

* start is included, stop is excluded
* every part is optional: default start = beginning, stop = end, step = 1
* a slice returns a new object of the same type

```python
cities = ["Riga", "Daugavpils", "Liepaja", "Jelgava", "Jurmala", "Ventspils", "Rezekne"]

print(cities[1:4])     # elements 1, 2, 3
print(cities[:3])      # first three
print(cities[3:])      # from index 3 to the end
print(cities[-2:])     # last two
print(cities[::2])     # every second element
print(cities[::-1])    # reversed copy
print(cities[0:5:2])
```

```python
# The same syntax works on a tuple and on a string
rgb = (255, 128, 0)
print(rgb[0], rgb[1:])

course = "Programming for Data Analytics"
print(course[0], course[-1], course[:11])
```

### Lists: adding elements
| Method | What it does |
|---|---|
| `lst.append(x)` | adds **one** element to the end |
| `lst.insert(i, x)` | inserts `x` before position `i` |
| `lst.extend(seq)` | adds **all** elements of `seq` to the end |
| `lst + other` | builds a **new** list, originals unchanged |

```python
stops = ["Central Station", "Old Town", "University"]

stops.append("Airport")
print(stops)

stops.insert(1, "Market")
print(stops)

stops.extend(["Harbour", "Zoo"])
print(stops)

more = stops + ["Depot"]           # a NEW list, stops is unchanged
print(more)
print(stops)
```

```python
# Typical mistake: append vs extend
a = [1, 2, 3]
a.append([4, 5])
print(a, len(a))

b = [1, 2, 3]
b.extend([4, 5])
print(b, len(b))
```

## Lists: removing elements

| Method | Removes by | Returns |
|---|---|---|
| `lst.remove(x)` | **value** (first match); `ValueError` if missing | nothing |
| `lst.pop()` | position: the last element | the removed element |
| `lst.pop(i)` | position `i` | the removed element |
| `del lst[i]` | position | nothing |
| `lst.clear()` | everything | nothing |

```python
stops = ["Central Station", "Market", "Old Town", "University", "Airport", "Harbour", "Zoo"]

stops.remove("Market")
print(stops)

last = stops.pop()
print(last, stops)

second = stops.pop(1)
print(second, stops)

del stops[0]
print(stops)

stops.clear()
print(stops)

# ERROR: list.remove(x): x not in list
# [1, 2, 3].remove(10)
```

## Set operations: the full picture

| Operation | Method | Operator | Result |
|---|---|---|---|
| Union | `A.union(B)` | `A \| B` | in A **or** B |
| Intersection | `A.intersection(B)` | `A & B` | in A **and** B |
| Difference | `A.difference(B)` | `A - B` | in A, **not** in B |
| Symmetric difference | `A.symmetric_difference(B)` | `A ^ B` | in **exactly one** |

Comparisons: `A.issubset(B)`, `A.issuperset(B)`, `A.isdisjoint(B)`, `A == B`

Full table: https://wesmckinney.com/book/python-builtin.html#tbl-table_set_operations

```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7}

print("Union               :", A | B)
print("Intersection        :", A & B)
print("Difference A - B    :", A - B)
print("Difference B - A    :", B - A)
print("Symmetric difference:", A ^ B)

print({1, 2}.issubset(A), A.issuperset({1, 2}), A.isdisjoint({10, 20}))
print({1, 2, 3} == {3, 2, 1})     # order does not matter
```

## Two courses, two registration lists (as they came from the registration form).

* How many different students registered for the Python course?
* Who takes both courses?
* Who takes Python but not Statistics?
* Who takes exactly one of the two courses?
* Does python_course contain a duplicate registration? Print True or False.


```python
python_course = ["Anna", "Janis", "Marta", "Olegs", "Anna", "Liga"]
stats_course = ["Marta", "Peteris", "Liga", "Ilze", "Olegs"]

python_set = set(python_course)
stats_set = set(stats_course)

print("Different Python students:", len(python_set))

print("Both courses:", python_set & stats_set)

print("Python only:", python_set - stats_set)

print("Exactly one course:", python_set ^ stats_set)

has_duplicates = len(python_course) != len(python_set)
print("Duplicate registration:", has_duplicates)
```

## Dictionaries

A dictionary stores a collection of key-value pairs, where key and value are Python objects. Each key is associated with a value so that a value can be conveniently retrieved, inserted, modified, or deleted given a particular key.

```python
dict1 = {'Monday': 'Moon', 'Tuesday' : 'Mars', 'Wednesday' : 'Mercury', 'Thursday' : 'Jupiter',
         'Friday' : 'Venus','Saturday' : 'Saturn'}
print(dict1)
print(type(dict1))
print(dict1.keys())
print(dict1.values())
```

```python
dict1['Sunday'] = 'Sun'
print(dict1)
```

```python
dict1.pop('Monday')
print(dict1)
```

## Dictionaries: reading

| Expression | Result if the key is missing |
|---|---|
| `d[key]` | `KeyError` |
| `d.get(key)` | `None` |
| `d.get(key, default)` | `default` |
| `key in d` | `False` |

- `in` checks **keys**. For values: `x in d.values()`
- `len(d)` — the number of key-value pairs

```python
capitals = {"Latvia": "Riga", "Estonia": "Tallinn", "Lithuania": "Vilnius"}

print(capitals["Latvia"])
print(len(capitals))

# ERROR: KeyError: 'Poland'
# print(capitals["Poland"])

print(capitals.get("Poland"))
print(capitals.get("Poland", "unknown"))
print("Poland" in capitals)
print("Riga" in capitals)                   # False - Riga is a value, not a key
print("Riga" in capitals.values())
```

## Dictionaries: adding, updating, deleting
* d[key] = value — a new key adds a pair, an existing key replaces the value
*  d.update(other) — several pairs at once
* d.pop(key) — removes the pair and returns its value; del d[key] — just removes

```python
capitals["Finland"] = "Helsinki"       # new key
print(capitals)

capitals["Finland"] = "Helsingfors"    # existing key - the value is replaced
print(capitals)

capitals.update({"Sweden": "Stockholm", "Finland": "Helsinki"})
print(capitals)

removed = capitals.pop("Sweden")
print(removed)

del capitals["Finland"]
print(capitals)
```

### Hashability

* A dictionary (and a set) finds elements through a hash — a number computed from the object's content
* The hash must never change → only immutable objects are hashable
* Hashable: int, float, str, bool, tuple — not hashable: list, set, dict

```python
coords[[50.8476, 4.3572]] = 'Brussel'
print(coords)
```

```python
coords[{50.8476, 4.3572}] = 'Brussel'
print(coords)
```

```python
hash((50.8476, 4.3572))
```
