## Round 2 - the building blocks of Python

*Week 2: data types, control flow, data structures*

# A number is read from a file as `"42"`. Why can't you add it to a total straight away?

# It is text (str), not a number

The type decides what an operation means: `"4" + "2"` joins text. Convert first - `int("42")`, `float("3.5")`. Checking types is one of the first steps with any new data.

# `if`, `for`, `while` - say in a few words what each one is for.

# Decide · repeat for each · repeat until

`if` chooses a branch, `for` walks through a collection you already have, `while` repeats as long as a condition is true (you do not know how many times).

Which tool would you choose?  list, dict, tuple, set, if, for, while, def, import, with open, try / except, NumPy array



# You need to store a table - 8 students, each with a name, group and grade - using only core Python. Which structure?

# A list of dictionaries

One dictionary per row (`{"name": ..., "grade": ...}`), all rows in a list. It works - but it is clumsy for big tables. That is the gap pandas fills today.

Which tool would you choose?  list, dict, tuple, set, if, for, while, def, import, with open, try / except, NumPy array



# You must check whether a student ID is in a list of 1 million IDs - very often. List or set?

# set (or dict keys)

A set finds an element almost instantly; a list is searched one element after another. The right data structure is often the biggest speed-up.

# The file is closed automatically

## Round 4 - NumPy: data as arrays

*Week 4: NumPy, images as arrays*

# Why NumPy arrays instead of Python lists for numbers?

# Fast, compact, whole-array operations

One block of memory with one data type; `arr * 2` works on all elements at once (10-50× faster than a loop) and works in 2, 3 or more dimensions.

# The iris data is an array of shape `(150, 4)`. What does that tell you?

# 150 rows (flowers) × 4 columns (measurements)

`shape` is the first thing to check on any array - the same as asking "how many records and how many features?"

# `iris.mean(axis=0)` - one number in total, one per row, or one per column?

# One per column

`axis=0` walks down the rows and gives one result for each column: the average of each measurement. Statistics per feature is what you want most often.

# What does `iris[iris[:, 0] > 7]` do - in plain words?

# Keeps only the flowers with sepal length > 7

A comparison gives a True/False mask; the mask selects the rows. Filtering with a condition is the most used idea in data analysis - you will meet it again in pandas.

# For NumPy, what is a colour photo?

# A 3D array: height × width × 3 colour channels

Numbers 0-255 for red, green and blue. Cropping = slicing, flipping = reversing an axis, a filter = arithmetic on neighbouring pixels. A grayscale photo is 2D.

# And a short video?

# A 4D array: frames × height × width × channels

The same tools work - one more dimension is time. Different data (tables, images, sound, video) all become arrays of numbers.

# You slice a part of an array and change it. Why does the original change too?

# A slice is a view, not a copy

It shares the same memory - fast, but surprising. Use `.copy()` when the original must stay untouched.

## The bridge to today

What can a NumPy array **not** do well with a real table like the bike rentals?

# Names, mixed types, missing values

A NumPy array has no column names, one data type for everything and no comfortable
way to handle gaps. **pandas** adds exactly that - on top of NumPy.
