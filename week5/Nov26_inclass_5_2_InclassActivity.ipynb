# 5.3. In-class activity: Collect Data - Provide application - Do something meaningful - Share Findings

Module M-227-04: Programming for Data Analytics

Instructor: Nadežda Zeņina

## Timing: 1.5h

1. Teaming (2-3 students)
2. Task + Presenting the results

## Task

> **1. Data collection** - Organise data collection from the notebook (e.g., downloading of data from open sources) or URL. 
>
> **2. Business understanding** - Describe potential applications of the data set - how and for which purposes it can be used.
>
> **3. Loading the data set** - The data set can be heterogeneous and include numeric data, text fragments, images, etc. This is highly recommended to use NumPy, Pandas and/or pure Python data structures.
>
> **4. Meaningful analysis** - Do something meaningful with the data (indexing, filtering, calculation, aggregation) .
>
> **5. Show the result to the group**  - and say what it means


## What you may use - everything from weeks 1-5

| Week | Topic | Typical tools for today |

### Where to find data

**Search yourself (preferred) - choose a topic close to your work or hobby**

| Source | What is there |
|---|---|
| Latvian open data portal - `data.gov.lv` | transport (Rīgas Satiksme timetables, e-ticket validations), regions, budgets |
| Official statistics of Latvia - `stat.gov.lv` | population, wages, prices, tourism |
| Eurostat - `ec.europa.eu/eurostat` | the same indicators for all EU countries |
| Our World in Data - `ourworldindata.org` | health, energy, CO2, education - every chart has a *Download* button |
| Kaggle - `kaggle.com/datasets` | thousands of data sets on any topic (download needs an account → use a local file) |
